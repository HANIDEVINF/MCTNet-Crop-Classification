## Step 2 · Configuration

```python
import ee

YEAR      = 2021
STEP_DAYS = 10      # days per composite window
N_STEPS   = 36      # number of time steps (covers full year)
N_SAMPLES = 2000    # per class; 5 classes x 2000 = ~10 000 total (AR), 6 x 2000 = ~12 000 (CA)
TILE_SCALE = 8      # tileScale for sampleRegions — higher = less memory per tile

# Sentinel-2 band mapping: GEE name -> our output name
BAND_MAP = {
    'B2':  'Blue',
    'B3':  'Green',
    'B4':  'Red',
    'B5':  'RE1',
    'B6':  'RE2',
    'B7':  'RE3',
    'B8A': 'RE4',
    'B8':  'NIR',
    'B11': 'SWIR1',
    'B12': 'SWIR2',
}
S2_BANDS  = list(BAND_MAP.keys())     # GEE band IDs
OUT_BANDS = list(BAND_MAP.values())   # our output names

# CDL class mappings per state
CDL_CONFIG = {
    'Arkansas': {
        'from': [1, 2, 3, 5],
        'to':   [1, 2, 3, 4],   # Corn=1, Cotton=2, Rice=3, Soybeans=4, Other=0
        'n_classes': 5,
    },
    'California': {
        'from': [1, 2, 3, 5, 75],
        'to':   [1, 2, 3, 4,  5],  # + Almonds=5
        'n_classes': 6,
    },
}

print('Configuration loaded')
print(f'Bands: {OUT_BANDS}')
print(f'Time steps: t01 – t{N_STEPS:02d}')
```

---
## Step 3 · Helper Functions

```python
import ee

def mask_s2_clouds(image):
    """
    Cloud masking using the SCL (Scene Classification Layer) band.
    Removes clouds (9), cloud shadows (3), and cirrus (10).
    Scales reflectance to [0, 1] by dividing by 10000.
    """
    scl  = image.select('SCL')
    mask = (scl.neq(3)   # cloud shadow
              .And(scl.neq(8))   # cloud medium probability
              .And(scl.neq(9))   # cloud high probability
              .And(scl.neq(10))) # cirrus
    return image.updateMask(mask).divide(10000).select(S2_BANDS)


def make_composite(step_index, roi):
    """
    Build one 10-day median composite for the given 0-based step index.
    Returns an image with bands renamed as Blue_t01, Green_t01, etc.
    Uses explicit per-step filtering to avoid GEE memory limits.
    """
    start_date = ee.Date(f'{YEAR}-01-01')
    t0     = start_date.advance(step_index * STEP_DAYS, 'day')
    t1     = t0.advance(STEP_DAYS, 'day')
    step   = f't{step_index + 1:02d}'
    suffix = f'_{step}'

    s2 = (ee.ImageCollection('COPERNICUS/S2_SR_HARMONIZED')
            .filterDate(t0, t1)
            .filterBounds(roi)
            .select(S2_BANDS + ['SCL'])
            .map(mask_s2_clouds))

    composite = s2.median().select(S2_BANDS)
    renamed   = composite.rename([b + suffix for b in OUT_BANDS])
    return renamed


def build_stack(roi):
    """
    Build the full 36-step stack by explicitly chaining .addBands() calls.
    This avoids ee.List.map() and toBands() which exceed GEE memory limits.
    """
    print('Building composite stack (36 steps)...')
    stack = make_composite(0, roi)
    for i in range(1, N_STEPS):
        stack = stack.addBands(make_composite(i, roi))
        if (i + 1) % 6 == 0:
            print(f'  Added step {i+1}/{N_STEPS}')
    print('Stack built   ')
    return stack


def get_roi(state_name):
    """
    Get the full state boundary from TIGER dataset, simplified to reduce
    vertex count and avoid GEE memory errors during spatial operations.
    """
    states = ee.FeatureCollection('TIGER/2018/States')
    return (states
            .filter(ee.Filter.eq('NAME', state_name))
            .geometry()
            .simplify(1000))  # 1000m tolerance — critical for memory efficiency


def get_label_image(state_name):
    """Build the remapped CDL label image for a given state."""
    cfg  = CDL_CONFIG[state_name]
    cdl  = (ee.ImageCollection('USDA/NASS/CDL')
              .filter(ee.Filter.calendarRange(YEAR, YEAR, 'year'))
              .first()
              .select('cropland'))
    return cdl.remap(cfg['from'], cfg['to'], defaultValue=0).rename('label')


print('Helper functions defined')
```

## Step 4 · Export Function

```python
import ee

def export_state(state_name, dry_run=False):
    """
    Full pipeline for one state:
    1. Build ROI and label image
    2. Build 36-step Sentinel-2 stack
    3. Stratified sampling
    4. Export to Google Drive

    Args:
        state_name: 'Arkansas' or 'California'
        dry_run: if True, skip export (just validate the stack)
    """
    print(f'\n{"="*55}')
    print(f'Processing: {state_name}')
    print('='*55)

    # 1. Region of interest
    roi = get_roi(state_name)
    print(f'ROI loaded   ')

    # 2. CDL labels
    label = get_label_image(state_name)
    print(f'CDL labels loaded   ')

    # 3. Build Sentinel-2 stack
    stack = build_stack(roi)

    # 4. Combine stack + label for sampling
    features = stack.addBands(label)

    # 5. Stratified random sampling
    cfg     = CDL_CONFIG[state_name]
    samples = features.stratifiedSample(
        numPoints  = N_SAMPLES,           # per class
        classBand  = 'label',
        region     = roi,
        scale      = 10,                  # 10m — Sentinel-2 native resolution
        tileScale  = TILE_SCALE,          # reduces memory per tile
        seed       = 42,
        geometries = True,                # exports .geo column with lat/lon
    )
    print(f'Sampling configured ({N_SAMPLES} per class)')

    if dry_run:
        print('Dry run — skipping export.')
        return

    # 6. Export to Google Drive
    task = ee.batch.Export.table.toDrive(
        collection  = samples,
        description = f'MCTNet_{state_name}_{YEAR}',
        fileFormat  = 'CSV',
        folder      = 'MCTNet_Data',
    )
    task.start()
    print(f'Export task started')
    print(f'→ Check https://code.earthengine.google.com/ Tasks tab')
    print(f'→ Output: Google Drive / MCTNet_Data / MCTNet_{state_name}_{YEAR}.csv')
    return task


print('Export function defined')
```

## Step 6 · Monitor Task Status

```python
import time

def check_tasks():
    """Print status of all active GEE export tasks."""
    tasks = ee.batch.Task.list()
    mctnet_tasks = [t for t in tasks if 'MCTNet' in t.config.get('description', '')]
    if not mctnet_tasks:
        print('No MCTNet tasks found.')
        return
    for t in mctnet_tasks:
        status = t.status()
        print(f"{status['description']:45s} | {status['state']:12s} | {status.get('error_message', '')}")

check_tasks()
```

```python
# Poll every 60 seconds until all tasks finish (optional)
import time

def wait_for_tasks(poll_seconds=60):
    while True:
        tasks  = ee.batch.Task.list()
        active = [t for t in tasks
                  if 'MCTNet' in t.config.get('description', '')
                  and t.status()['state'] in ('READY', 'RUNNING')]
        if not active:
            print('All MCTNet tasks finished')
            check_tasks()
            break
        print(f'  {len(active)} task(s) still running... checking again in {poll_seconds}s')
        time.sleep(poll_seconds)

# Uncomment to enable polling:
# wait_for_tasks()
```

## Step 7 · Preprocessing (after downloading CSVs from Google Drive)

Once the GEE exports finish:
1. Download `MCTNet_Arkansas_2021.csv` and `MCTNet_California_2021.csv` from **Google Drive → MCTNet_Data/**
2. Place them in `./data/raw/`
3. Run the cells below to clean and validate them


```python
import pandas as pd
import numpy as np
import glob, os, re

RAW_DIR    = './data/raw/'
OUTPUT_DIR = './data/processed/'
os.makedirs(OUTPUT_DIR, exist_ok=True)

STATES  = ['Arkansas', 'California']
N_STEPS = 36
BANDS   = ['Blue','Green','Red','RE1','RE2','RE3','RE4','NIR','SWIR1','SWIR2']

print('Preprocessing config loaded')
```

```python
def preprocess(state):
    print(f'\n{"="*50}')
    print(f'Processing {state}')
    print('='*50)

    # 1. Load (merge shards if GEE split the export)
    files = sorted(glob.glob(os.path.join(RAW_DIR, f'MCTNet_{state}_{YEAR}*.csv')))
    if not files:
        raise FileNotFoundError(f'No CSV found for {state} in {RAW_DIR}')
    df = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)
    print(f'Loaded {len(files)} file(s) → {len(df)} rows × {len(df.columns)} cols')

    # 2. Validate band columns
    expected = [f'{b}_t{t:02d}' for b in BANDS for t in range(1, N_STEPS+1)]
    missing  = [c for c in expected if c not in df.columns]
    if missing:
        print(f'  ⚠ WARNING: {len(missing)} band columns missing — check GEE export')
    else:
        print(f'  All {len(expected)} band×time columns present')

    # 3. Fill NaN (cloudy time steps) with forward/backward fill per band
    for band in BANDS:
        cols = [f'{band}_t{t:02d}' for t in range(1, N_STEPS+1)]
        present = [c for c in cols if c in df.columns]
        df[present] = df[present].ffill(axis=1).bfill(axis=1)

    # 4. Verify reflectance scale [0, 1]
    for band in BANDS:
        cols = [f'{band}_t{t:02d}' for t in range(1, N_STEPS+1) if f'{band}_t{t:02d}' in df.columns]
        mx = df[cols].max().max()
        if mx > 1.5:
            print(f'  ⚠ WARNING: {band} max = {mx:.3f} (expected ≤ 1.0, check divide by 10000)')

    # 5. Label distribution
    print('  Label distribution:')
    print('  ' + df['label'].value_counts().sort_index().to_string().replace('\n', '\n  '))

    # 6. Save
    out = os.path.join(OUTPUT_DIR, f'test_MCTNet_{state}_{YEAR}_fixed.csv')
    df.to_csv(out, index=False)
    print(f'  Saved → {out}')
    return df

for state in STATES:
    preprocess(state)

print('\nDone   ')
```
